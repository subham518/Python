```python
import pandas as pd
```

```python
# read csv file
df_cars = pd.read_csv('Car_sales.csv')
```

```python
# selecing columns
df_cars = df_cars[['Manufacturer', 'Sales_in_thousands',
         'Vehicle_type', 'Price_in_thousands',
         'Engine_size', 'Horsepower', 'Fuel_capacity']]
```

```python
# show dataframe
df_cars
```

# agg()

```python
# calculate the sum of columns
df_cars.agg('sum')
```

```python
# calculate the mean of numeric columns
df_cars.mean(numeric_only=True)
```

```python
# calculate the mean and count values
# df_cars.select_dtypes(include='number').apply(['mean', 'count'])
df_cars.select_dtypes(include='number').agg(['mean', 'sum'])
```

```python
# Different aggregations per column: calculate the sum and mean of 'Sale_in_thousands' and sum and max of 'Price_in_thousands'
df_cars.agg({'Sales_in_thousands':['sum','mean'],
            'Price_in_thousands':['sum','max']})
```

```python
# aggregate over the columns
df_cars[['Sales_in_thousands', 'Price_in_thousands']].agg('sum', axis = 1)
```

```python
# equivalent (NaN behaves differently)
df_cars['Sales_in_thousands'] + df_cars['Price_in_thousands']
```

```python
# aggregate different functions and rename the index of the resulting Dataframe
df_cars.agg(x = ('Sales_in_thousands', 'sum'),
            y = ('Price_in_thousands','sum'))
```

# The Split-Apply-Combine Strategy

```python
# split data into separate groups
passenger_filter = df_cars['Vehicle_type'] == 'Passenger'
car_filter = df_cars['Vehicle_type'] == 'Car'

# df_cars['Vehicle_type'].value_counts()
```

```python
# apply an operation
passenger_avg = df_cars[passenger_filter]['Sales_in_thousands'].mean()
car_avg = df_cars[car_filter]['Sales_in_thousands'].mean()
```

```python
# combine results
pd.DataFrame({'Vehicle_type':['Car','Passenger'],
             'Sales_in_thousands':[car_avg, passenger_avg]})
# pd.DataFrame({'Vehicle_type':['Car','Passenger'],
#              'Sales_in_thousands':[car_avg, passenger_avg]}).set_index('Vehicle_type')
```

# Groupby() and agg()

```python
# find the minimum and maximum values on each column
# df_cars.groupby('Vehicle_type').min()
# df_cars.groupby('Vehicle_type').max()

df_cars.groupby('Vehicle_type').agg(['min', 'max'])
```

```python
# set a name for aggregated column
# find the min Engine_size and max Horsepower
df_cars.groupby('Vehicle_type').agg(min_engine_size = ('Engine_size', 'min'),
                                   max_horsepower = ('Horsepower', 'max'))
```

```python
# group by 'Manufacturer' and calculate the sum of Sales_in_thousands and mean of Price_in_thousands
df_cars.groupby('Manufacturer').agg(sales_sum = ('Sales_in_thousands', 'sum'),
                                   price_mean = ('Price_in_thousands', 'mean'))
```

```python
# df_cars.groupby('Manufacturer').sum()['Sales_in_thousands']
df_cars.groupby('Manufacturer')['Price_in_thousands'].mean()
```

# Lambda

Apply your own function

```python
# get actual sales and price values
# df_cars.groupby('Manufacturer')[['Sales_in_thousands',
#                                  'Price_in_thousands']].sum().apply(lambda x:x*1000)
df_cars.groupby('Manufacturer').sum().apply(lambda x:x*1000)[['Sales_in_thousands',
                                 'Price_in_thousands']]
```

```python
# subtract values from the mean of each group
# df_cars.groupby('Manufacturer').apply(lambda x:x-x.mean(numeric_only=True), include_groups=False)

#or:
# Separate the numeric columns first, but keep the group column
# numeric_cols = df_cars.select_dtypes(include='number').columns

# Apply the operation only to those numeric columns
# df_cars.groupby('Manufacturer')[numeric_cols].apply(lambda x: x - x.mean())

# modern way:
# 1. Select your numeric columns
numeric_cols = df_cars.select_dtypes(include='number').columns

# 2. Subtract the group mean directly
df_cars[numeric_cols] - df_cars.groupby('Manufacturer')[numeric_cols].transform('mean')
```
