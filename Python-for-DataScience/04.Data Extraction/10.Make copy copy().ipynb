```python
import pandas as pd
```

```python
#reads csv file
df = pd.read_csv('players_20.csv')
# set index
df.set_index('short_name', inplace=True)
# select column
df = df[['long_name', 'age', 'dob', 'height_cm', 'weight_kg', 'nationality', 'club']]
```

```python
# show dataframe
df
```

# copy()

## deep = True

```python
# deep = True by default (Modifications to the data or indices of the copy will not be reflected)
df_copy = df.copy()
```

```python
# update value in original  dataframe
df.loc['L. Messi', 'height_cm'] = 180
```

```python
# copy vs original dataframe
df
df_copy
```

## deep = False

```python
# deep = False (Any changes to the data of the original will be reflected)
df_copy = df.copy(deep=False)
```

```python
# update value in original dataframe
df.loc['Cristiano Ronaldo', 'height_cm'] = 180
```

```python
# shallow copy vs original dadtaframe
df_copy
df
```

## Simple assignment

```python
# make a copy
df_new_copy = df
```

```python
# update value in original dataframe
df.loc['Neymar Jr', 'height_cm'] = 190
```

```python
# copy vs original dataframe
df
df_new_copy
```
